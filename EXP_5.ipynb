```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create a sample diabetes dataset
data = {
    'Pregnancies': [2, 0, 5, 1, 3, 4, 0, 6],
    'Glucose': [120, 85, 150, 95, 130, 145, 90, 160],
    'BloodPressure': [70, 65, 80, 60, 75, 85, 68, 90],
    'BMI': [25.5, 22.0, 32.1, 24.0, 28.5, 30.2, 23.5, 35.0],
    'Age': [35, 22, 45, 28, 32, 50, 25, 55],
    'Outcome': [1, 0, 1, 0, 1, 1, 0, 1]
}

diabetes_df = pd.DataFrame(data)

# Display first five rows
print("First Five Rows:")
print(diabetes_df.head())

# Display dataset information
print("\nDataset Information:")
diabetes_df.info()

# Check missing values
print("\nMissing Values:")
print(diabetes_df.isnull().sum())

# Statistical summary
print("\nStatistical Summary:")
print(diabetes_df.describe())

# Display correlation matrix
print("\nCorrelation Matrix:")
print(diabetes_df.corr())

# Visualize the data
sns.heatmap(diabetes_df.corr(), annot=True, cmap='coolwarm')
plt.title("Diabetes Dataset Correlation")
plt.show()

# Plot glucose distribution
plt.hist(diabetes_df['Glucose'], bins=5)
plt.title("Glucose Distribution")
plt.xlabel("Glucose")
plt.ylabel("Frequency")
plt.show()
```
